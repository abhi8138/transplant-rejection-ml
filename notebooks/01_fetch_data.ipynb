Fetching the data matrix from GEO Accession: GSE150059

```python
from pathlib import Path
from urllib.request import urlretrieve

url = "https://ftp.ncbi.nlm.nih.gov/geo/series/GSE150nnn/GSE150059/matrix/GSE150059_series_matrix.txt.gz"
dest = Path("../data/GSE150059_series_matrix.txt.gz")   # notebook lives in notebooks/
dest.parent.mkdir(exist_ok=True)

if not dest.exists():          # skip if already downloaded
    urlretrieve(url, dest)
print(dest, dest.stat().st_size / 1e6, "MB")
```
